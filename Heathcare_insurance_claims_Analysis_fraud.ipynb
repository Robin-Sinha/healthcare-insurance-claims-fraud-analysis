# EDA

Step 1 — Fraud Distribution

```python
df['Is_Fraud'].value_counts()
```

```python
df['Is_Fraud'].value_counts(normalize=True) * 100
```

Step 2 — Diagnosis Analysis

```python
df.groupby('Diagnosis_Code')['Is_Fraud'].mean().mul(100).round(2)
```

```python
df.groupby('Diagnosis_Code')['Is_Fraud'].agg(['count', 'sum'])
```

Step 3 — Provider Analysis

```python
df.groupby('Provider_ID')['Is_Fraud'].mean().mul(100).round(2).sort_values(ascending=False)
```

```python
df.groupby('Provider_ID')['Is_Fraud'].agg(['count', 'sum']).sort_values('sum', ascending=False).head(10)
```

```python
df.groupby('Provider_ID')['Claim_Amount'].mean().sort_values(ascending=False).head(10)
```

Step 4 — Claim Amount Analysis

```python
df.groupby('Is_Fraud')['Claim_Amount'].mean()
```

```python
df.groupby('Is_Fraud').agg(
    Avg_Claim_Amount=('Claim_Amount', 'mean'),
    Avg_Approved_Amount=('Approved_Amount', 'mean')
)
```

Step 5 — Final Findings

```python
print("Total Claims:", len(df))
print("Fraud-labelled Claims:", (df['Is_Fraud'] == 1).sum())
print("Non-fraud-labelled Claims:", (df['Is_Fraud'] == 0).sum())
```
